## Car names column

```python
def repeated_values(column):
    values = sorted(column.unique())
    values_set = set()
    j = 0
    for i in range(1, len(values)):
        if values[i].lower().strip() == values[j].lower().strip():
            if values[i] not in values_set:
                values_set.add(values[i])
        i = i + 1
        j = j + 1
    return values_set
            
```

```python
values_set = repeated_values(df["Cars Names"])
print(values_set)
```

# Conclusion
All 11 columns are stored as text (`str`) and require cleaning before any
numeric analysis or modeling is possible. Common issues across columns
include:

- Units embedded in values (`cc`, `hp`, `Nm`, `km/h`, `sec`)
- Currency symbols and thousands separators (`$`, `,`)
- Ranges and multi-value entries that need to be resolved into a single
  number (ex., `100 - 150 hp`, `365 Nm / 480 Nm`)
- Inconsistent casing and duplicate categories (`Nissan` vs `NISSAN`)
- Corrupted/unknown characters, likely from encoding issues (`?`)
- A few domain-specific edge cases: hybrid/electric combined values,
  `2+2` seating configuration, and outlier prices for rare or
  concept-only vehicles.
